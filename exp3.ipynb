### 5. Activity 2 — First-Principles Implementation (Without Built-in Functions)

```python
# Implementation of Statistical Functions from Scratch (Without built-in statistical functions)

def custom_mean(data):
    '''Calculates arithmetic mean: sum(x) / N'''
    total = 0.0
    count = 0
    for x in data:
        total += x
        count += 1
    return total / count if count > 0 else float('nan')

def custom_median(data):
    '''Calculates median by sorting and finding middle element(s)'''
    s_data = sorted(data)
    n = len(s_data)
    if n == 0:
        return float('nan')
    mid = n // 2
    if n % 2 == 1:
        return float(s_data[mid])
    else:
        return (s_data[mid - 1] + s_data[mid]) / 2.0

def custom_mode(data):
    '''Calculates mode using a pure dictionary frequency counter'''
    freq_map = {}
    for x in data:
        freq_map[x] = freq_map.get(x, 0) + 1
    max_freq = 0
    for val, count in freq_map.items():
        if count > max_freq:
            max_freq = count
    modes = [val for val, count in freq_map.items() if count == max_freq]
    return modes[0] if modes else float('nan'), max_freq

def custom_range(data):
    '''Calculates range: max - min using single-pass min/max iteration'''
    if not data:
        return float('nan'), float('nan'), float('nan')
    x_min = data[0]
    x_max = data[0]
    for x in data[1:]:
        if x < x_min:
            x_min = x
        if x > x_max:
            x_max = x
    return x_max - x_min, x_min, x_max

def custom_variance(data, ddof=1):
    '''Calculates sample variance: sum((x - mean)^2) / (N - ddof)'''
    n = len(data)
    if n <= ddof:
        return float('nan')
    mu = custom_mean(data)
    sum_sq_diff = 0.0
    for x in data:
        sum_sq_diff += (x - mu) ** 2
    return sum_sq_diff / (n - ddof)

def custom_std_dev(data, ddof=1):
    '''Calculates sample standard deviation using Newton-Raphson square root'''
    var_val = custom_variance(data, ddof=ddof)
    if var_val < 0 or var_val != var_val:
        return float('nan')
    if var_val == 0:
        return 0.0
    # Newton-Raphson square root
    root = var_val / 2.0
    for _ in range(50):
        root = 0.5 * (root + var_val / root)
    return root

def custom_percentile(sorted_data, p):
    '''Calculates percentile via standard linear interpolation'''
    n = len(sorted_data)
    if n == 0:
        return float('nan')
    idx = p * (n - 1)
    low = int(idx)
    high = min(low + 1, n - 1)
    weight = idx - low
    return (1.0 - weight) * sorted_data[low] + weight * sorted_data[high]

def custom_quartiles_and_iqr(data):
    '''Calculates Q1, Q2, Q3, and IQR'''
    s_data = sorted(data)
    q1 = custom_percentile(s_data, 0.25)
    q2 = custom_percentile(s_data, 0.50)
    q3 = custom_percentile(s_data, 0.75)
    iqr = q3 - q1
    return q1, q2, q3, iqr

def custom_cv(data, ddof=1):
    '''Calculates Coefficient of Variation: (s / mean) * 100'''
    mu = custom_mean(data)
    s = custom_std_dev(data, ddof=ddof)
    return (s / mu) * 100.0 if mu != 0 else float('nan')

print("Custom first-principles statistical module successfully compiled.")
```

**Inference:** Implementing the statistical routines from first principles highlights foundational algorithmic principles:
1. **Computational Complexity:** The arithmetic mean, range, and variance execute in linear time $O(n)$ with $O(1)$ auxiliary space.
2. **Sorting Overhead:** Positional metrics (median, quartiles, and IQR) require sorting, which operates in $O(n \log n)$ time.
3. **Bessel's Correction:** Dividing by $N - 1$ instead of $N$ corrects for the negative bias inherent in estimating the population variance from a sample.

### 7. Activity 3 — Discretization and Binning using Sturges' Rule

```python
# 1. Parameter Derivation via Sturges' Rule
n = len(df['heat_index'].dropna())
k = int(np.ceil(1 + np.log2(n)))
x_min = df['heat_index'].min()
x_max = df['heat_index'].max()
range_val = x_max - x_min
width = range_val / k

print(f"Total Observations (n)  : {n}")
print(f"Minimum Value (x_min)   : {x_min:.4f} °C")
print(f"Maximum Value (x_max)   : {x_max:.4f} °C")
print(f"Total Range (R)         : {range_val:.4f} °C")
print(f"Class Intervals (k)     : {k} (via Sturges' Rule: ceil(1 + log2(n)))")
print(f"Class Width (W or h)    : {width:.4f} °C")

# 2. Continuous Discretization via pd.cut
df['heat_index_interval'] = pd.cut(df['heat_index'], bins=k)

# 3. Frequency Distribution Table Construction
freq_series = df['heat_index_interval'].value_counts(sort=False)
freq_table = freq_series.reset_index()
freq_table.columns = ['Class Interval (°C)', 'Frequency (fi)']
freq_table['Relative Freq (%)'] = (freq_table['Frequency (fi)'] / n) * 100
freq_table['Cumulative Freq (cf)'] = freq_table['Frequency (fi)'].cumsum()
freq_table['Midpoint (mi)'] = [round(x.mid, 4) for x in freq_table['Class Interval (°C)']]
freq_table['fi * mi'] = freq_table['Frequency (fi)'] * freq_table['Midpoint (mi)']

print("\n--- Complete Frequency Distribution Table (heat_index) ---")
display(freq_table[['Class Interval (°C)', 'Midpoint (mi)', 'Frequency (fi)', 'Relative Freq (%)', 'Cumulative Freq (cf)', 'fi * mi']])
```

**Inference:** Discretizing the continuous `heat_index` variable using Sturges' Rule produces **$k = 18$ equal-width classes** ($W \approx 3.86^\circ\text{C}$). The frequency distribution reveals that the modal class is **(22.817, 26.679]°C**, accommodating **17,141 records** (**16.17%** of the entire dataset). Approximately **69.3%** of all historical observations are concentrated between **22.82°C and 45.99°C**, representing moderate-to-hot Indian climatic baselines.

### 8. Activity 3 — Mathematical Formulation of Grouped Data Statistics (heat_index)

```python
# Extract interval arrays and frequency vectors
intervals = freq_series.index
frequencies = freq_series.values.astype(float)
midpoints = freq_table['Midpoint (mi)'].values.astype(float)
N = frequencies.sum()
cum_freq = np.cumsum(frequencies)

# 1. Grouped Mean: x̄ = Σ(fi * mi) / N
grouped_mean = np.sum(frequencies * midpoints) / N

# Helper for Grouped Percentiles: L + [ (target - cf_prev) / f_group ] * h
def get_grouped_percentile(p):
    target = p * N
    idx = int(np.searchsorted(cum_freq, target))
    L = intervals[idx].left
    h = intervals[idx].length
    cf_prev = cum_freq[idx - 1] if idx > 0 else 0.0
    f = frequencies[idx]
    return L + ((target - cf_prev) / f) * h, idx, L, h, cf_prev, f

# 2. Grouped Median (50th percentile)
grouped_median, med_idx, med_L, med_h, med_cf_prev, med_f = get_grouped_percentile(0.50)

# 3. Grouped Quartiles and IQR
grouped_q1, q1_idx, q1_L, q1_h, q1_cf_prev, q1_f = get_grouped_percentile(0.25)
grouped_q3, q3_idx, q3_L, q3_h, q3_cf_prev, q3_f = get_grouped_percentile(0.75)
grouped_iqr = grouped_q3 - grouped_q1

# 4. Grouped Mode: L + [ (f1 - f0) / (2f1 - f0 - f2) ] * h
modal_idx = int(np.argmax(frequencies))
L_mode = intervals[modal_idx].left
h_mode = intervals[modal_idx].length
f1 = frequencies[modal_idx]
f0 = frequencies[modal_idx - 1] if modal_idx > 0 else 0.0
f2 = frequencies[modal_idx + 1] if modal_idx < len(frequencies) - 1 else 0.0
denom = (2 * f1 - f0 - f2)
grouped_mode = L_mode + ((f1 - f0) / denom) * h_mode if denom != 0 else midpoints[modal_idx]

# 5. Grouped Variance and Standard Deviation: s² = Σ fi*(mi - x̄)² / (N - 1)
grouped_variance = np.sum(frequencies * (midpoints - grouped_mean) ** 2) / (N - 1)
grouped_std = np.sqrt(grouped_variance)

print("--- Grouped Data Statistical Parameters ---")
print(f"Total Sample Size (N)             : {int(N)}")
print(f"Grouped Mean (x̄_grp)              : {grouped_mean:.4f} °C")
print(f"Median Class                      : {intervals[med_idx]} (L={med_L:.3f}, h={med_h:.3f}, f={int(med_f)}, cf_prev={int(med_cf_prev)})")
print(f"Grouped Median                    : {grouped_median:.4f} °C")
print(f"Modal Class                       : {intervals[modal_idx]} (L={L_mode:.3f}, f0={int(f0)}, f1={int(f1)}, f2={int(f2)})")
print(f"Grouped Mode                      : {grouped_mode:.4f} °C")
print(f"Grouped Sample Variance (s²_grp)  : {grouped_variance:.4f}")
print(f"Grouped Sample Std Dev (s_grp)    : {grouped_std:.4f} °C")
print(f"Grouped Q1 (25th percentile)      : {grouped_q1:.4f} °C")
print(f"Grouped Q3 (75th percentile)      : {grouped_q3:.4f} °C")
print(f"Grouped IQR (Q3 - Q1)             : {grouped_iqr:.4f} °C")
```

**Inference:** Grouped calculations illustrate the mathematical mechanism of approximating continuous distributions from binned tables:
1. **Grouped Median (34.86°C):** Falls within the class interval **(30.542, 34.405]°C**, where linear interpolation between lower boundary $L$ and upper boundary accounts for cumulative density up to $N/2 = 52,990$.
2. **Grouped Mode (25.88°C):** Captures the apex of probability density in the third class via weighted modal class differences ($f_1 - f_0$ and $f_1 - f_2$), accurately identifying peak empirical density.

### 10. Activity 3 — Multi-Attribute Grouped Frequency & Statistics (humidity & peak_met_mw)

```python
def compute_grouped_attribute_stats(series, attribute_name, unit):
    n_attr = len(series.dropna())
    k_attr = int(np.ceil(1 + np.log2(n_attr)))
    x_min_attr = series.min()
    x_max_attr = series.max()
    w_attr = (x_max_attr - x_min_attr) / k_attr
    
    # Categorical cut
    interval_series = pd.cut(series, bins=k_attr)
    f_series = interval_series.value_counts(sort=False)
    
    intervs = f_series.index
    f_vals = f_series.values.astype(float)
    mids = np.array([iv.mid for iv in intervs])
    N_tot = f_vals.sum()
    c_freq = np.cumsum(f_vals)
    
    # Grouped Mean
    g_mean = np.sum(f_vals * mids) / N_tot
    
    # Grouped Median
    def get_pct(p):
        tgt = p * N_tot
        idx = int(np.searchsorted(c_freq, tgt))
        L = intervs[idx].left
        h = intervs[idx].length
        cf_prev = c_freq[idx - 1] if idx > 0 else 0.0
        return L + ((tgt - cf_prev) / f_vals[idx]) * h
    
    g_med = get_pct(0.50)
    g_q1 = get_pct(0.25)
    g_q3 = get_pct(0.75)
    g_iqr = g_q3 - g_q1
    
    # Grouped Mode
    m_idx = int(np.argmax(f_vals))
    L_m = intervs[m_idx].left
    h_m = intervs[m_idx].length
    f1_m = f_vals[m_idx]
    f0_m = f_vals[m_idx - 1] if m_idx > 0 else 0.0
    f2_m = f_vals[m_idx + 1] if m_idx < len(f_vals) - 1 else 0.0
    den = 2 * f1_m - f0_m - f2_m
    g_mode = L_m + ((f1_m - f0_m) / den) * h_m if den != 0 else mids[m_idx]
    
    # Grouped Variance & Std Dev
    g_var = np.sum(f_vals * (mids - g_mean) ** 2) / (N_tot - 1)
    g_std = np.sqrt(g_var)
    
    return {
        'Attribute': f"{attribute_name} ({unit})",
        'Grouped Mean': round(g_mean, 2),
        'Grouped Median': round(g_med, 2),
        'Grouped Mode': round(g_mode, 2),
        'Grouped Std Dev': round(g_std, 2),
        'Grouped IQR': round(g_iqr, 2),
        'Raw Mean': round(series.mean(), 2),
        'Raw Median': round(series.median(), 2),
        'Raw Std Dev': round(series.std(), 2),
        'Raw IQR': round(series.quantile(0.75) - series.quantile(0.25), 2)
    }

stats_humid = compute_grouped_attribute_stats(df['humidity'], 'humidity', '%')
stats_peak = compute_grouped_attribute_stats(df['peak_met_mw'], 'peak_met_mw', 'MW')

grouped_multi_summary = pd.DataFrame([stats_humid, stats_peak])
print("--- Grouped Statistics Summary: `humidity` and `peak_met_mw` ---")
display(grouped_multi_summary)
```

**Inference:** The grouped estimation holds robustly across disparate attribute domains:
1. **Humidity (%):** Grouped mean (**66.52%**) and grouped median (**68.21%**) match the raw continuous values (**66.50%** and **68.21%**), preserving the left-skewed nature of relative humidity.
2. **Peak Met Demand (MW):** Grouped mean (**5,671.27 MW**) closely tracks raw mean (**5,672.09 MW**), and grouped median (**4,570.62 MW**) mirrors raw median (**4,557.00 MW**), capturing the substantial positive skew and long tail of electricity demand.

### 12. Dispersion Bounds — Empirical Rule and Chebyshev's Inequality Verification

```python
def evaluate_dispersion_bounds(series, attr_name):
    mu = series.mean()
    sigma = series.std()
    N_total = len(series)
    
    results = []
    for k_val in [1, 2, 3]:
        lower_b = mu - k_val * sigma
        upper_b = mu + k_val * sigma
        count_within = ((series >= lower_b) & (series <= upper_b)).sum()
        empirical_pct = (count_within / N_total) * 100
        chebyshev_min = (1.0 - (1.0 / (k_val ** 2))) * 100 if k_val > 1 else 0.0
        gaussian_rule = {1: 68.27, 2: 95.45, 3: 99.73}[k_val]
        
        results.append({
            'k (Std Deviations)': f"k = {k_val}",
            'Interval [μ - kσ, μ + kσ]': f"[{lower_b:.2f}, {upper_b:.2f}]",
            'Empirical Proportion (%)': round(empirical_pct, 2),
            'Chebyshev Lower Bound (%)': f"≥ {chebyshev_min:.1f}%",
            'Normal Distribution Rule (%)': f"{gaussian_rule:.2f}%",
            'Chebyshev Satisfied?': 'YES' if empirical_pct >= chebyshev_min else 'NO'
        })
    return pd.DataFrame(results)

print("--- Dispersion Bounds: `heat_index` ---")
display(evaluate_dispersion_bounds(df['heat_index'], 'heat_index'))

print("\n--- Dispersion Bounds: `total_demand_mw` ---")
display(evaluate_dispersion_bounds(df['total_demand_mw'], 'total_demand_mw'))
```

**Inference:** The empirical verification confirms mathematical inequality bounds:
1. **Chebyshev's Inequality:** Validated with 100% compliance across all attributes. At $k = 2$, at least $75\%$ is guaranteed, whereas empirical coverage is **95.84%** for `heat_index` and **95.27%** for `total_demand_mw`. At $k = 3$, coverage exceeds **98.8%**, well above the $88.9\%$ Chebyshev minimum.
2. **Empirical Rule Compliance:** `heat_index` tracks the Gaussian 68-95-99.7 rule closely (**67.82%**, **95.84%**, **99.82%**), reflecting its near-normal bell distribution. In contrast, `total_demand_mw` exhibits higher concentration within $1\sigma$ (**82.14%**), characteristic of a right-skewed distribution with a long tail.

### Q1. What are the various applications of central tendency and variability of data?

Measures of central tendency (Mean, Median, Mode) and variability (Range, Variance, Standard Deviation, IQR, Coefficient of Variation) form the core pillars of quantitative data science, predictive modeling, and engineering applications:

1. **Exploratory Data Analysis (EDA) and Distribution Profiling:**
   - Comparing Mean versus Median reveals distribution asymmetry: $Mean > Median$ flags positive (right) skewness, while $Mean < Median$ flags negative (left) skewness.
   - The Coefficient of Variation ($CV = \frac{\sigma}{\mu} \times 100\%$) allows dimensionless comparison of dispersion across variables with different physical units (e.g., comparing temperature in °C with power demand in MW).

2. **Feature Engineering, Normalization, and Standardization:**
   - **Z-Score Standardization:** Transforms features to zero mean and unit variance:
     $$z = \frac{x - \mu}{\sigma}$$
     essential for distance-based machine learning algorithms (k-NN, SVM, PCA).
   - **Robust Scaling:** Uses Median and IQR for scaling heavily skewed or outlier-contaminated datasets:
     $$x_{\text{robust}} = \frac{x - \text{Median}}{\text{IQR}}$$

3. **Statistical Anomaly and Outlier Detection:**
   - **Tukey's IQR Fences:** Points outside $[Q_1 - 1.5 \times \text{IQR}, \; Q_3 + 1.5 \times \text{IQR}]$ are isolated as outliers without making parametric Gaussian assumptions.
   - **Three-Sigma ($3\sigma$) Rule:** In automated monitoring, values beyond $\mu \pm 3\sigma$ flag critical equipment failures or grid abnormalities.

4. **Power Grid Engineering and Infrastructure Capacity Planning:**
   - The arithmetic mean establishes baseline base-load generation requirements.
   - Variability measures (Std Dev and Peak-to-Mean Ratio) determine the required spinning reserve capacity and battery storage margins to prevent blackouts during peak load swings.

5. **Financial Risk Assessment and Investment Management:**
   - Expected portfolio return is quantified by the Mean, while investment volatility/risk is quantified by the Standard Deviation.
   - The **Sharpe Ratio** ($S = \frac{R_p - R_f}{\sigma_p}$) evaluates risk-adjusted return.

6. **Statistical Quality Control and Six Sigma Engineering:**
   - Industrial processes rely on $\bar{X}$ (mean) and $S$ (standard deviation) control charts to ensure manufactured tolerances remain within $6\sigma$ specifications ($C_p, C_{pk} \ge 1.33$).


```python
# Demonstration of Application: Tukey's IQR Outlier Detection and Robust Scaling

# Case 1: Tukey's IQR Outlier Detection on Electrical Grid Deficits (peak_unmet_mw)
unmet_series = df['peak_unmet_mw']
q1_u = unmet_series.quantile(0.25)
q3_u = unmet_series.quantile(0.75)
iqr_u = q3_u - q1_u

lower_fence = q1_u - 1.5 * iqr_u
upper_fence = q3_u + 1.5 * iqr_u
outliers = unmet_series[(unmet_series < lower_fence) | (unmet_series > upper_fence)]

print("--- Application 1: Outlier Detection via Tukey's IQR Fences ---")
print(f"Q1: {q1_u:.2f} MW, Q3: {q3_u:.2f} MW, IQR: {iqr_u:.2f} MW")
print(f"Lower Fence: {lower_fence:.2f} MW | Upper Fence: {upper_fence:.2f} MW")
print(f"Total Grid Deficit Outlier Records: {len(outliers)} ({len(outliers)/len(unmet_series)*100:.2f}%)")
print(f"Top 5 Extreme Deficit Events (MW):\n{outliers.sort_values(ascending=False).head(5).values}")

# Case 2: Z-Score vs. Robust Scaling on Skewed Total Demand
sample_demand = df['total_demand_mw'].head(1000)
z_scaled = (sample_demand - sample_demand.mean()) / sample_demand.std()
robust_scaled = (sample_demand - sample_demand.median()) / (sample_demand.quantile(0.75) - sample_demand.quantile(0.25))

scaling_comparison = pd.DataFrame({
    'Raw Demand (MW)': sample_demand.head(5).values,
    'Z-Score Scaled (μ=0, σ=1)': z_scaled.head(5).round(4).values,
    'Robust Scaled (Med=0, IQR=1)': robust_scaled.head(5).round(4).values
})
print("\n--- Application 2: Feature Scaling Comparison ---")
display(scaling_comparison)
```

### Q2. What are the limitations of Central Tendency and variability of Data?

While central tendency and variability provide concise numerical summaries, relying exclusively on them introduces critical analytical pitfalls:

1. **Extreme Sensitivity of Mean and Variance to Outliers (Zero Breakdown Point):**
   - The arithmetic mean and variance have a **breakdown point of 0%**, meaning that even a single corrupted entry or severe outlier can arbitrarily distort both metrics.
   - In contrast, the median has a breakdown point of **50%**, resisting contamination up to half the sample.

2. **Concealment of Multimodal and Non-Symmetric Distributions:**
   - A single measure of central tendency implies a unimodal clustering center. In bimodal or multimodal distributions, the mean often falls into an antimode (a low-probability valley where few or no actual observations exist).
   - Datasets with identical means and standard deviations can have drastically different geometric structures (**Anscombe's Quartet** and the **Datasaurus Dozen**).

3. **Ambiguity and Fragility of the Mode in Continuous Data:**
   - In continuous, real-valued distributions (such as temperature or power load with floating-point decimals), exact value repetitions rarely occur, rendering raw mode meaningless unless the data is discretized into class bins.

4. **Inappropriateness for Non-Interval Scales (Categorical / Ordinal Data):**
   - Computing the arithmetic mean on ordinal ratings (e.g., customer satisfaction ranks 1 to 5) falsely presumes equidistant intervals between subjective ranks.
   - For nominal variables (e.g., state names, machine IDs), mean and variance are undefined; only the mode is mathematically valid.

5. **Loss of Temporal, Spatial, and Sequential Dynamics:**
   - Central tendency and dispersion collapse multi-dimensional time series into static scalars, erasing cyclical trends, diurnal rhythms, autocorrelation, and seasonal phase shifts.

6. **Simpson's Paradox and Aggregation Bias:**
   - Aggregating disparate sub-populations can cause overall central tendencies to show a trend that is completely reversed when analyzing individual subgroups.

| Statistical Metric | Core Strength | Critical Limitation | Recommended Alternative / Remedy |
| :--- | :--- | :--- | :--- |
| **Arithmetic Mean (x̄)** | Mathematically tractable, utilizes all values | Extreme vulnerability to outliers; zero breakdown point | Use **Median** or **Trimmed / Winsorized Mean** |
| **Median** | Highly robust against outliers (50% breakdown point) | Lacks algebraic tractability; insensitive to distribution tails | Pair with **IQR** and quartile plots |
| **Mode** | Applicable to nominal/categorical data | Fragile/undefined in continuous real-valued arrays | Use **Discretized Modal Class** or Kernel Density Estimation |
| **Variance / Std Dev (s², s)** | Quantifies spread around the mean; fundamental to probability theory | Squared deviations over-penalize tail outliers | Use **Median Absolute Deviation (MAD)** or **IQR** |
| **Range** | Simple to compute and understand | Entirely determined by extreme two values; ignores inner distribution | Use **Interquartile Range (IQR)** or **Percentile Ranges** |
| **Coefficient of Variation (CV)** | Dimensionless, scale-invariant relative spread | Undefined or unstable when mean approaches zero | Use **Quartile Coefficient of Dispersion** |


```python
# Demonstration of Limitations: Outlier Distortion and Multimodal Failure

# 1. Outlier Sensitivity Test (Breakdown Point Demonstration)
base_demand = np.array([4500.0, 4700.0, 4800.0, 4900.0, 5000.0, 5100.0, 5200.0, 5300.0, 5500.0])
contaminated_demand = np.append(base_demand, 500000.0) # Introduce extreme grid outlier (500,000 MW)

outlier_impact = pd.DataFrame({
    'Metric': ['Mean (MW)', 'Median (MW)', 'Std Dev (MW)', 'IQR (MW)'],
    'Clean Sample': [base_demand.mean(), np.median(base_demand), base_demand.std(ddof=1), np.percentile(base_demand, 75) - np.percentile(base_demand, 25)],
    'Contaminated (+500k Outlier)': [contaminated_demand.mean(), np.median(contaminated_demand), contaminated_demand.std(ddof=1), np.percentile(contaminated_demand, 75) - np.percentile(contaminated_demand, 25)]
})
outlier_impact['Percentage Distortion (%)'] = ((outlier_impact['Contaminated (+500k Outlier)'] - outlier_impact['Clean Sample']) / outlier_impact['Clean Sample']) * 100

print("--- Limitation 1: Catastrophic Outlier Distortion on Mean vs. Median ---")
display(outlier_impact.round(2))

# 2. Multimodal Distribution Fallacy Demonstration
# Construct a bimodal distribution with modes at 20 and 80
bimodal_sample = np.concatenate([np.random.normal(20, 3, 500), np.random.normal(80, 3, 500)])
bm_mean = bimodal_sample.mean()
bm_median = np.median(bimodal_sample)
near_mean_count = np.sum((bimodal_sample >= bm_mean - 5) & (bimodal_sample <= bm_mean + 5))

print("\n--- Limitation 2: Multimodal Distribution Fallacy ---")
print(f"Bimodal Dataset: True Clusters at ~20 and ~80")
print(f"Calculated Arithmetic Mean : {bm_mean:.2f}")
print(f"Calculated Median          : {bm_median:.2f}")
print(f"Observations within ±5 units of Mean: {near_mean_count} out of {len(bimodal_sample)} ({near_mean_count/len(bimodal_sample)*100:.1f}%)")
print("Conclusion: The mean lies in an empty valley where virtually no actual observations exist!")
```
